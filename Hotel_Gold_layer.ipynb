```python
df_trim = spark.read.table("hotel.silver.hotel_bookings_cleaned")
display(df_trim)
```

```python
from pyspark.sql.functions import count, sum

df_agg = df_trim.groupBy("check_in_date") \
    .agg(
        count("*").alias("total_booking"),
        sum("total_amount").alias("total_revenue")
    ) \
    .withColumnRenamed("check_in_date", "date") \
    .orderBy("date")

display(df_agg)
```

```python
from pyspark.sql.functions import count, sum, month, year

df_monthly = df_trim.withColumn("year", year("check_in_date")) \
    .withColumn("month", month("check_in_date")) \
    .groupBy("year", "month") \
    .agg(
        count("*").alias("total_booking"),
        sum("total_amount").alias("total_revenue")
    ) \
    .orderBy("year", "month")

display(df_monthly)
```

```python
from pyspark.sql.functions import sum

df_city_revenue = df_trim.groupBy("hotel_city") \
    .agg(
        sum("total_amount").alias("total_revenue")
    ) \
    .orderBy("total_revenue", ascending=False)

display(df_city_revenue)
```

```python
from pyspark.sql.functions import count, sum

df_roomtype = df_trim.groupBy("room_type") \
    .agg(
        count("*").alias("total_booking"),
        sum("total_amount").alias("total_revenue")
    ) \
    .orderBy("total_revenue", ascending=False)

display(df_roomtype)
```

```python
# booking_status
from pyspark.sql.functions import count, sum

df_booking_status = df_trim.groupBy("booking_status") \
    .agg(
        count("*").alias("total_booking"),
        sum("total_amount").alias("total_revenue")
    ) \
    .orderBy("total_revenue", ascending=False)

display(df_booking_status)
```

```python
# df_agg
df_agg.write.format("delta").mode("overwrite").saveAsTable("hotel.gold.hotel_date_wise")
```

```python
# df_monthly
df_monthly.write.format("delta").mode("overwrite").saveAsTable("hotel.gold.hotel_monthly")
# df_city_revenue
df_city_revenue.write.format("delta").mode("overwrite").saveAsTable("hotel.gold.hotel_city_wise")
# df_roomtype
df_roomtype.write.format("delta").mode("overwrite").saveAsTable("hotel.gold.hotel_room_type_wise")
# df_booking_status
df_booking_status.write.format("delta").mode("overwrite").saveAsTable("hotel.gold.hotel_booking_status")
```
